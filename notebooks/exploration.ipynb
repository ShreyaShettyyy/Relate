# Relate Recommendation Engine Exploration
This notebook demonstrates Market Basket Analysis using the Apriori algorithm, Association Rule Mining, and Multi-Rail Recommendation Generation.

```python
import pandas as pd
from recommender import RecommenderSystem

df_trans = pd.read_csv('transactions.csv')
df_prod = pd.read_csv('products.csv')
print(f'Loaded {len(df_trans)} transactions and {len(df_prod)} catalog products.')
df_trans.head()
```

```python
# Initialize Recommender and Train Apriori Model
rec = RecommenderSystem(df_trans, df_prod)
rec.train_market_basket(min_support=0.001, min_threshold=0.1)
print(f'Frequent itemsets: {len(rec.frequent_itemsets)}')
print(f'Association rules mined: {len(rec.rules)}')
```

```python
# Top Association Rules by Lift and Confidence
top_rules = rec.rules.sort_values('lift', ascending=False).head(10)
top_rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']]
```

```python
# Inspect the 4 Recommendation Rails for Smartphone
item = 'iPhone 13 Pro'
print('1. Frequently Bought Together:', rec.get_frequently_bought_together(item))
print('2. Recommended Accessories:', rec.get_recommended_accessories(item))
print('3. Similar Products:', rec.get_similar_products(item))
print('4. You May Also Like:', rec.get_you_may_also_like(item))
```

```python
# Inspect the 4 Recommendation Rails for Laptop
item = 'Apple MacBook Pro 14 Inch Space Grey'
print('1. Frequently Bought Together:', rec.get_frequently_bought_together(item))
print('2. Recommended Accessories:', rec.get_recommended_accessories(item))
print('3. Similar Products:', rec.get_similar_products(item))
print('4. You May Also Like:', rec.get_you_may_also_like(item))
```
